# Run TRACE and QE-ORP side by side.

```
from trace.env.world import default_world
from trace.sim.engine import simulate
from trace.baselines.trace_protocol import TRACEProtocol
from trace.baselines.qe_orp import QEORPProtocol

world = default_world(seed=0, n_nodes=20)
r_trace = simulate(TRACEProtocol(), world, duration_h=0.2, seed=0)
r_qe = simulate(QEORPProtocol(), world, duration_h=0.2, seed=0)
print('TRACE  PDR =', round(r_trace.pdr, 3),
      ' E/bit =', round(r_trace.energy_per_bit_nj))
print('QE-ORP PDR =', round(r_qe.pdr, 3),
      ' E/bit =', round(r_qe.energy_per_bit_nj))
```
